```python
plt.plot(df["Salary"], color = "red", marker = "*",linestyle = ":",linewidth = "2")
plt.grid()
plt.show()
```

```python
# Histogram
plt.hist(df["Salary"], bins = 5, color = "green")
plt.show()
```

```python
plt.boxplot(df["Salary"])
plt.show()
```

```python
plt.boxplot(df["Salary"])
plt.show()
```

```python
df.drop(index = 20, inplace=True)
df.shape
```

```python
# Univariate: Categorical
```

```python
df["Dept"] = ['HR','IT','Finance','HT','Finance','IT','HR','Finance','IT']
```

```python
df.head()
```

```python
df.info()
```

```python
# Pie Chart:

count = df["Dept"].value_counts()
count
```

```python
plt.pie(count, labels = count.index, autopct = "%1.1f", explode=[0, 0.1, 0.2, 0.3])
plt.axis("equal")
plt.show()
```

```python
#Count plot
plt.bar(count.index, count)
plt.show()
```

```python
# Bivariate - numerical - numerical
df["Age"] = [22, 25, 36, 42, 27, 22, 21, 28, 32]
```

```python
df.head()
```

```python
df.shape
```

```python
# Scatter Plot:

plt.scatter(df["Age"],df["Salary"],color = "orange")
```

```python
sorted_age = df.sort_values("Age")
```

```python
# Line plot:
plt.plot(sorted_age["Age"],df["Salary"])
```

```python
# Bar chart:
plt.bar(sorted_age["Age"], df["Salary"],align = "center")
```

```python
# Bivariate: Numerical - Categorical
```

```python
hr_sal = df[df["Dept"] == "HR"]["Salary"]
hr_sal
```

```python
it_sal = df[df["Dept"] == "IT"]["Salary"]
it_sal
```

```python
finance_sal = df[df["Dept"] == "Finance"]["Salary"]
finance_sal
```

```python
plt.boxplot([hr_sal, it_sal, finance_sal], labels = ["HR", "IT", "Finance"])
plt.legend()
plt.show()
```

```python
# Pie Chart
salary_by_dept = df.groupby("Dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels = salary_by_dept, autopct="%1.1f",shadow=True,explode=[0.1,0.1,0.1,0.1])
plt.axis("equal")
```

```python
# bar plot
hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fin_mean = sum(finance_sal)/len(finance_sal)
```

```python
plt.bar(["HR","Finance","IT"],[hr_mean, it_mean, fin_mean], color = ['green','black','red'])
```

```python
# Multivariate Analysis: 3 numerical columns
```

```python
df["Experience"] = [1,4,3,2,1,6,2,1,5]
```

```python
df.head()
```

```python
# Bubble Plot:
plt.scatter(df["Age"],df["Salary"], s = df["Experience"]*20, color = "orange", edgecolors="black")
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.show()
```

```python
# 2 numerical and 1 categorical column:
```

```python
color = {"HR":"yellow","IT":"blue","Finance":"orange"}
for dept, color in color.items():
    df_dept = df[df["Dept"] == dept]
    plt.scatter(df_dept["Age"],df_dept["Salary"],c = color, label = dept)
    
plt.legend()
```

```python
# Object Oriented API
```

```python
fig, axs = plt.subplots(1,3, figsize = (10,4))

# Line Plot

axs[0].plot(sorted_age["Age"],df["Salary"], color = "red", marker = "*", linewidth = "2", markersize = 2)
axs[0].grid()
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

# Histogram

axs[1].hist(df["Salary"],bins = 5, color = "green")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

# Scatter Plot
axs[2].boxplot(df["Salary"])
axs[2].set_title("BoxPlot")
axs[2].set_xlabel("Salary")
```

```python
data2 = {
    'Year':[2020,2001,2009,2010],
    'Sales':[100, 150, 200, 250],
    'Profits':[20, 30, 40, 50],
    'Expenses':[80, 120, 160, 200]
}
```

```python
df2 = pd.DataFrame(data2)
```

```python
plt.plot(df2["Year"],df2["Sales"], label = "Sales")
plt.plot(df2["Year"],df2["Profits"], label = "Profits")
plt.plot(df2["Year"],df2["Expenses"], label = "Expenses")

plt.title("Financial Analysis")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
```

```python
# 3D Plot

ax = plt.axes(projection = "3d")
ax.set_xlabel("Age")
ax.set_ylabel("Salary")
ax.set_zlabel("Experience")
ax.scatter(df["Age"], df["Salary"], df["Experience"])
```
